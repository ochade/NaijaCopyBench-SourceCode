```python
import json, re, math
import statistics as st
from pathlib import Path

ROOT = Path.cwd().parent
m = json.loads((ROOT/"data/derived/section_map.json").read_text(encoding="utf-8"))
secs = {int(k): v for k, v in m.items() if k != "SCHEDULE"}
print(f"loaded {len(secs)} sections")

PARTS = {
    "I Subsistence": range(1,20),   "II Exceptions": range(20,28),
    "III Ownership": range(28,36),  "IV Infringement": range(36,44),
    "V Offences": range(44,48),     "VI Anti-piracy": range(48,54),
    "VII Online": range(54,63),     "VIII Performers": range(63,74),
    "IX Folklore": range(74,77),    "X Commission": range(77,91),
    "XI Financial": range(91,96),   "XII Miscellaneous": range(96,110),
}
```

```python
print("Part                    sections")
for name, rng in PARTS.items():
    n = sum(1 for s in secs if s in rng)
    print(f"{name:22} {n:2}  {'█'*n}")
```

```python
# Cell 3 — length distribution
lengths = {n: len(s["text"].split()) for n, s in secs.items()}
vals = sorted(lengths.values())
print(f"words/section — min {min(vals)}, median {int(st.median(vals))}, "
      f"mean {int(st.mean(vals))}, max {max(vals)}")
print(f"sections >400 words (RAG split candidates): {sum(1 for v in vals if v>400)}")
longest = sorted(lengths.items(), key=lambda x:-x[1])[:5]
print("longest:", [(n, secs[n]['title'][:28], w) for n,w in longest])
```

```python
# Cell 4 — Herdan's C
full = " ".join(s["text"] for s in secs.values()).lower()
tokens = re.findall(r"[a-z]+", full)
V, N = len(set(tokens)), len(tokens)
herdan_c = math.log(V) / math.log(N)
print(f"tokens N={N:,}  vocabulary V={V:,}")
print(f"Herdan's C = {herdan_c:.3f}  (report WITH N; comparators ~0.70 on larger corpora)")
```
